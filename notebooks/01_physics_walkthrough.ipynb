JCA properties and rigid-/air-backed TMM responses using the package implementation.

```python
import numpy as np
import matplotlib.pyplot as plt
from acoustic_coating.validation import representative_material
from acoustic_coating.tmm import rigid_backed_response
f = np.linspace(100, 5000, 1000)
material = representative_material()
response = rigid_backed_response(material, f)
plt.plot(f, response['absorption'])
plt.xlabel('Frequency [Hz]'); plt.ylabel('Absorption [-]'); plt.ylim(0, 1); plt.grid(alpha=.25);
```
